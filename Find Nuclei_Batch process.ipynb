```python
data_root_path = Path(os.path.expanduser("~")) / "Documents/Python Scripts/Infer-subc-2D"

in_data_path = data_root_path /"Z:/Maria Clara/2_Lab data/1_Multispectral data/2023/112023_MSi08-L_Neuronal Differentiation - iPSCs-hNGN2/Deconvolved iN day21 images 082024/tiff/more pics"
out_data_path = data_root_path/"Z:/Maria Clara/2_Lab data/9_Napari/inferSubC iNday21 more pics"
im_type = ".tiff"

img_file_list = list_image_files(in_data_path,im_type)
```

BATCH CODE

```python
suffix="nuc"
img_file_list = list_image_files(in_data_path, im_type)
for img_f in img_file_list:
    img_data, meta_dict = read_czi_image(img_f)
    agg = make_aggregate(img_in=img_data,
                     weight_ch0 = 0,
                     weight_ch1= 0,
                     weight_ch2= 1,
                     weight_ch3= 1,
                     weight_ch4= 1,
                     weight_ch5= 5,
                     weight_ch6= 20,
                     weight_ch7= 0,
                     weight_ch9= 0,
                     rescale= True)
    sns = scale_and_smooth(img_in=agg, 
                       median_size= 0, 
                       gauss_sigma= 4, 
                       slice_by_slice= True)
    mot = masked_object_thresh(structure_img_smooth= sns,
                           global_method='tri',
                           cutoff_size=200,
                           local_adjust=0.3)
    fafls = fill_and_filter_linear_size(img=mot,
                                    hole_min=0,
                                    hole_max=0,
                                    min_size=5,
                                    method= "3D",
                                    connectivity= 1)
    infc = infer_nuclei_fromcytoplasm(cytoplasm_mask=fafls,
                                  nuc_min_width=50,
                                  nuc_max_width=100,
                                  fill_filter_method="3D",
                                  small_obj_width=20)
   
    export_inferred_organelle(infc, suffix, meta_dict=meta_dict, out_data_path=out_data_path)
```
